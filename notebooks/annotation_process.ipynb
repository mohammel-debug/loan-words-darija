# Data Loading ([Darija Open Dataset](https://github.com/darija-open-dataset/dataset))

```python
import pandas as pd

url = "https://github.com/darija-open-dataset/dataset/blob/main/sentences/sentences.csv?raw=true"
dataset = pd.read_csv(url)

display(dataset.shape)
dataset.head()
```
